## 1. Mathematical & Theoretical Framework

### 1.1 Risk-Adjusted Performance Metrics

#### Annualized Log Returns and Volatility:
$$\mu_{\text{ann}} = 252 \cdot \frac{1}{T} \sum_{t=1}^{T} r_t, \quad \sigma_{\text{ann}} = \sqrt{252} \cdot \sqrt{\frac{1}{T-1} \sum_{t=1}^{T} (r_t - \bar{r})^2}$$

#### Sharpe & Sortino Ratios:
$$\text{Sharpe} = \frac{\mu_{\text{ann}} - R_f}{\sigma_{\text{ann}}}, \quad \text{Sortino} = \frac{\mu_{\text{ann}} - R_f}{\sqrt{252} \cdot \sqrt{\frac{1}{T} \sum_{t=1}^T \min(0, r_t)^2}}$$

#### Value-at-Risk (VaR) & Expected Shortfall (CVaR):
$$\text{VaR}_{\alpha} = -F_r^{-1}(1-\alpha), \quad \text{CVaR}_{\alpha} = -\mathbb{E}[r_t \mid r_t \le -\text{VaR}_{\alpha}]$$

### 1.2 Fibonacci Harmonic Geometry
Based on the divine proportion $\phi = \frac{1 + \sqrt{5}}{2} \approx 1.6180339887$:
- Key Retracement Ratios: $1 - \frac{1}{\phi} = 0.382$, $\frac{1}{\phi} = 0.618$ (**Golden Pocket**), $\phi - 1 = 0.618$, $\sqrt{\phi^{-1}} = 0.786$.

### 1.3 Position Sizing: Kelly Criterion & Volatility Targeting
$$f^* = \frac{p \cdot b - (1-p)}{b} \times 0.5, \quad w_{\text{vol}} = \min\left(0.20, \frac{\sigma_{\text{target}}}{\sigma_{\text{asset}}}\right)$$

## 4. Visualizing Fibonacci Golden Retracement Geometry

```python
fibo = thesis.fibonacci_framework
plt.figure(figsize=(12, 6))
plt.plot(asset_df.index, asset_df['Close'], label=f'{ticker} Close Price', color='#2563eb', lw=1.5)

levels = [
    ('Swing High', fibo.swing_high, '#ef4444'),
    ('Fib 23.6%', fibo.level_236, '#f97316'),
    ('Fib 38.2%', fibo.level_382, '#eab308'),
    ('Fib 50.0%', fibo.level_500, '#06b6d4'),
    ('Fib 61.8% (Golden Pocket)', fibo.level_618, '#10b981'),
    ('Fib 78.6%', fibo.level_786, '#8b5cf6'),
    ('Swing Low', fibo.swing_low, '#22c55e'),
]

for label, price, col in levels:
    plt.axhline(price, linestyle='--', color=col, alpha=0.7, label=f'{label}: ${price:.2f}')

plt.title(f'{ticker} Price Action & Fibonacci Retracement Levels ({fibo.trend_direction})', fontsize=14, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.grid(True, alpha=0.3)
plt.legend(loc='upper left', bbox_to_anchor=(1.01, 1), fontsize=9)
plt.tight_layout()
plt.show()
```
